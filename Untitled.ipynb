```python
feature = "RevolvingUtilizationOfUnsecuredLines"

# Learn bin boundaries from training data only
_, bin_edges = pd.qcut(
    X_train[feature],
    q=10,
    retbins=True,
    duplicates="drop"
)

# Extend boundaries to cover future observations
bin_edges[0] = -np.inf
bin_edges[-1] = np.inf

# Apply the same boundaries to training and validation
train_bins = pd.cut(
    X_train[feature],
    bins=bin_edges,
    include_lowest=True
).astype("string").fillna("Missing")

val_bins = pd.cut(
    X_val[feature],
    bins=bin_edges,
    include_lowest=True
).astype("string").fillna("Missing")

print(train_bins.value_counts())
```

```python
woe_data = pd.DataFrame({
    "Bin": train_bins,
    "Target": y_train
})

woe_table = woe_data.groupby(
    "Bin",
    observed=False
)["Target"].agg(
    Total="count",
    Bad="sum"
)

woe_table["Good"] = (
    woe_table["Total"] - woe_table["Bad"]
)

display(woe_table)
```

```python
# Total good and bad borrowers
total_good = (y_train == 0).sum()
total_bad = (y_train == 1).sum()

# Number of bins
n_bins = len(woe_table)

# Good and bad distributions
woe_table["Good_Distribution"] = (
    (woe_table["Good"] + 0.5) /
    (total_good + 0.5 * n_bins)
)

woe_table["Bad_Distribution"] = (
    (woe_table["Bad"] + 0.5) /
    (total_bad + 0.5 * n_bins)
)

# Weight of Evidence
woe_table["WoE"] = np.log(
    woe_table["Good_Distribution"] /
    woe_table["Bad_Distribution"]
)

# Information Value contribution
woe_table["IV_Contribution"] = (
    woe_table["Good_Distribution"] -
    woe_table["Bad_Distribution"]
) * woe_table["WoE"]

# Total Information Value
iv = woe_table["IV_Contribution"].sum()

print("Information Value:", round(iv, 4))

display(woe_table)
```

```python


import pandas as pd
import numpy as np

late_columns = [
    'NumberOfTime30-59DaysPastDueNotWorse',
    'NumberOfTime60-89DaysPastDueNotWorse',
    'NumberOfTimes90DaysLate'
]

def calculate_iv(feature, X_train, y_train):

    data = pd.DataFrame({
        'Feature': X_train[feature],
        'Target': y_train
    })

    # Handle special payment codes
    if feature in late_columns:
        data['Feature'] = data['Feature'].replace(
            [96, 98], np.nan
        )

    # Numeric feature binning
    if data['Feature'].nunique(dropna=True) > 10:

        _, edges = pd.qcut(
            data['Feature'],
            q=10,
            retbins=True,
            duplicates='drop'
        )

        edges[0] = -np.inf
        edges[-1] = np.inf

        data['Bin'] = pd.cut(
            data['Feature'],
            bins=edges,
            include_lowest=True
        ).astype('string').fillna('Missing')

    else:
        data['Bin'] = (
            data['Feature'].astype('string').fillna('Missing')
        )

    # Calculate Good and Bad
    table = data.groupby(
        'Bin', observed=False
    )['Target'].agg(
        Total='count',
        Bad='sum'
    )

    table['Good'] = table['Total'] - table['Bad']

    # Smoothing
    n = len(table)
    total_good = (y_train == 0).sum()
    total_bad = (y_train == 1).sum()

    table['Good_Dist'] = (
        (table['Good'] + 0.5) /
        (total_good + 0.5 * n)
    )

    table['Bad_Dist'] = (
        (table['Bad'] + 0.5) /
        (total_bad + 0.5 * n)
    )

    table['WoE'] = np.log(
        table['Good_Dist'] / table['Bad_Dist']
    )

    table['IV_Contribution'] = (
        (table['Good_Dist'] - table['Bad_Dist'])
        * table['WoE']
    )

    return table['IV_Contribution'].sum()


# Calculate IV for all features
iv_results = []

for feature in X_train.columns:

    iv = calculate_iv(
        feature,
        X_train,
        y_train
    )

    iv_results.append({
        'Feature': feature,
        'IV': iv
    })

iv_df = pd.DataFrame(iv_results)

iv_df = iv_df.sort_values(
    by='IV',
    ascending=False
).reset_index(drop=True)

display(iv_df.round(4))
```

```python
selected_features = [
    'RevolvingUtilizationOfUnsecuredLines',
    'age',
    'DebtRatio',
    'MonthlyIncome',
    'NumberOfOpenCreditLinesAndLoans',
    'NumberRealEstateLoansOrLines',
    'NumberOfDependents',
    'NumberOfTime30-59DaysPastDueNotWorse',
    'NumberOfTime60-89DaysPastDueNotWorse',
    'NumberOfTimes90DaysLate',
    'MonthlyIncome_Missing'
]

X_train_selected = X_train[selected_features].copy()
X_val_selected = X_val[selected_features].copy()

print(X_train_selected.shape)
print(X_val_selected.shape)
```

```python

from sklearn.metrics import roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import average_precision_score

woe_maps = {}
bin_edges_map = {}
X_train_woe = pd.DataFrame(index=X_train_selected.index)
X_val_woe = pd.DataFrame(index=X_val_selected.index)

for col in selected_features:

    train_col = X_train_selected[col].copy()
    val_col = X_val_selected[col].copy()

    # Handle special late-payment values
    if col in late_columns:
        train_col = train_col.replace([96, 98], np.nan)
        val_col = val_col.replace([96, 98], np.nan)

    # Learn bins using training data only
    if train_col.nunique(dropna=True) > 10:

        _, edges = pd.qcut(
            train_col,
            q=10,
            retbins=True,
            duplicates='drop'
        )

        edges[0] = -np.inf
        edges[-1] = np.inf

        train_bins = pd.cut(
            train_col, bins=edges,
            include_lowest=True
        )

        val_bins = pd.cut(
            val_col, bins=edges,
            include_lowest=True
        )

        bin_edges_map[col] = edges

    else:
        train_bins = train_col.astype('string')
        val_bins = val_col.astype('string')

    train_bins = train_bins.astype(object)
    val_bins = val_bins.astype(object)

    train_bins = train_bins.where(
        pd.notna(train_bins), 'Missing'
    )

    val_bins = val_bins.where(
        pd.notna(val_bins), 'Missing'
    )

    temp = pd.DataFrame({
        'Bin': train_bins,
        'Target': y_train
    })

    table = temp.groupby('Bin')['Target'].agg(
        Total='count',
        Bad='sum'
    )

    table['Good'] = table['Total'] - table['Bad']

    n = len(table)
    total_good = (y_train == 0).sum()
    total_bad = (y_train == 1).sum()

    good_dist = (
        (table['Good'] + 0.5) /
        (total_good + 0.5 * n)
    )

    bad_dist = (
        (table['Bad'] + 0.5) /
        (total_bad + 0.5 * n)
    )

    woe = np.log(good_dist / bad_dist)

    woe_maps[col] = woe.to_dict()

    # Transform both datasets using training WoE
    X_train_woe[col] = train_bins.map(
        woe_maps[col]
    ).fillna(0)

    X_val_woe[col] = val_bins.map(
        woe_maps[col]
    ).fillna(0)

print("Training WoE shape:", X_train_woe.shape)
print("Validation WoE shape:", X_val_woe.shape)
print("WoE transformation completed.")
```

```python

model = LogisticRegression(
    max_iter=2000,
    class_weight='balanced',
    random_state=42
)

model.fit(X_train_woe, y_train)

# Predictions
y_pred = model.predict(X_val_woe)
y_prob = model.predict_proba(X_val_woe)[:, 1]

# Evaluation
print("ROC-AUC:", roc_auc_score(y_val, y_prob))

print(
    "PR-AUC:",
    average_precision_score(y_val, y_prob)
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_val, y_pred))

print("\nClassification Report:")
print(classification_report(y_val, y_pred))
```

```python

from sklearn.metrics import precision_recall_curve
import pandas as pd
import numpy as np

precision, recall, thresholds = precision_recall_curve(
    y_val, y_prob
)

threshold_results = pd.DataFrame({
    "Threshold": thresholds,
    "Precision": precision[:-1],
    "Recall": recall[:-1]
})

threshold_results["F1_Score"] = (
    2 * threshold_results["Precision"] *
    threshold_results["Recall"] /
    (
        threshold_results["Precision"] +
        threshold_results["Recall"] + 1e-10
    )
)

# Best threshold based on F1
best_result = threshold_results.loc[
    threshold_results["F1_Score"].idxmax()
]

print("Best threshold:")
display(best_result.to_frame())

# Compare selected thresholds
display(
    threshold_results[
        threshold_results["Threshold"].isin(
            [0.20, 0.30, 0.40, 0.50]
        )
    ].round(3)
)
```

```python

from sklearn.calibration import CalibratedClassifierCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, average_precision_score

base_model = LogisticRegression(
    max_iter=2000,
    class_weight='balanced',
    random_state=42
)

calibrated_model = CalibratedClassifierCV(
    estimator=base_model,
    method='sigmoid',
    cv=5
)

calibrated_model.fit(X_train_woe, y_train)

# Calibrated delinquency probabilities
calibrated_prob = calibrated_model.predict_proba(
    X_val_woe
)[:, 1]

print(
    "Calibrated ROC-AUC:",
    roc_auc_score(y_val, calibrated_prob)
)

print(
    "Calibrated PR-AUC:",
    average_precision_score(y_val, calibrated_prob)
)
```

```python
default_probability = calibrated_model.predict_proba(
    X_val_woe
)[:, 1]

# Avoid log(0)
default_probability = np.clip(
    default_probability, 1e-6, 1 - 1e-6
)

# Scorecard parameters
base_score = 600
base_odds = 20
PDO = 50

factor = PDO / np.log(2)

offset = base_score - (
    factor * np.log(base_odds)
)

# Calculate good-to-bad odds
good_bad_odds = (
    (1 - default_probability) /
    default_probability
)

# Calculate credit score
credit_score = (
    offset + factor * np.log(good_bad_odds)
)

# Keep scores within display range
credit_score = np.clip(
    credit_score, 300, 850
)

# Create borrower-level scorecard
scorecard = pd.DataFrame({
    "Actual_Delinquency": y_val.values,
    "Default_Probability": default_probability,
    "Credit_Score": credit_score
})

scorecard["Credit_Score"] = (
    scorecard["Credit_Score"].round(0).astype(int)
)

print("Credit scorecard generated!")

display(scorecard.head(10))
```

```python
scorecard["Risk_Band"] = pd.qcut(
    scorecard["Default_Probability"],
    q=3,
    labels=["Low Risk", "Medium Risk", "High Risk"]
)

# Risk band summary
risk_summary = scorecard.groupby(
    "Risk_Band",
    observed=False
).agg(
    Borrowers=("Actual_Delinquency", "count"),
    Actual_Bad=("Actual_Delinquency", "sum"),
    Average_Score=("Credit_Score", "mean"),
    Average_Default_Probability=(
        "Default_Probability", "mean"
    )
)

risk_summary["Actual_Delinquency_Rate"] = (
    risk_summary["Actual_Bad"] /
    risk_summary["Borrowers"] * 100
)

display(risk_summary.round(3))
```
